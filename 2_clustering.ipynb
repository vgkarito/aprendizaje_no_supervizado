```python
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('./data/retailmax.csv')
```

```python
features = df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
type(scaled_features)
```

```python
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, n_init=10, random_state=42)
clusters = kmeans.fit_predict(scaled_features)
```

```python
clusters
```

```python
len(clusters)
```

```python
kmeans = KMeans(n_clusters=5, init='k-means++', max_iter=300, n_init=10, random_state=42)
clusters = kmeans.fit_predict(scaled_features)
```

```python
clusters
```

```python
kmeans = KMeans(n_clusters=4, init='k-means++', max_iter=300, n_init=10, random_state=42)
clusters = kmeans.fit_predict(scaled_features)
```

```python
clusters
```

```python
df
```

```python
df['clasificación'] = clusters
df
```

```python
df['clasificación'].value_counts()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='clasificación', palette='Set1')
plt.title('Clusters de clientes')
plt.show()
```

```python
 # Función para calcular la Suma de Cuadrados Dentro del Cluster (WCSS)
def calcular_wcss(datos):
    wcss = []
    for n in range(1, 11):
        kmeans = KMeans(n_clusters=n, init='k-means++', max_iter=300, n_init=10, random_state=42)
        kmeans.fit(datos)
        wcss.append(kmeans.inertia_)
    return wcss

# Calcular el WCSS para diferentes números de clusters
wcss = calcular_wcss(scaled_features)

# Graficar el método del codo
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.title('Método del codo')
plt.xlabel('Número de clusters')
plt.ylabel('WCSS (Inercia)')
plt.show()
```

```python
# Modelo final con el número óptimo de clusters
kmeans_final = KMeans(n_clusters=5, init='k-means++', max_iter=300, n_init=10, random_state=42)

clusters = kmeans_final.fit_predict(scaled_features)

# Agregar la clasificación al dataframe original
df['Cluster'] = clusters

df.head()
```

```python
df['Cluster'].value_counts().sort_index()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Cluster',
    palette='Set1',
    s=80
)

plt.title('Segmentación final de clientes (K-Means)')
plt.xlabel('Ingreso anual (k$)')
plt.ylabel('Spending Score')
plt.legend(title='Cluster')
plt.show()
```
